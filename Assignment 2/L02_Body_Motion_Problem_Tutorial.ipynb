1. **E1 · The Gap & Whole-Course Judgment**
   - Eva says "Serve me a drink." A video shows a robot executing a smooth trajectory, placing the mug on the table, but a photograph reveals milk spilled across the counter and an open fridge left behind. Explain why this nominal end-state success does **not** constitute a solution to the Body Motion Problem, and specify two whole-course constraints that were violated.

Solution: Reaching the final state does not necessarily mean that the task was completed successfully. Although the robot placed the mug on the table, the actions performed throughout the task must also satisfy the relevant constraints.

Two constraints were violated:

No spillage: The milk should be transferred into the mug without spilling. The spilled milk indicates that the pouring action was not properly controlled.

Fridge closed after use: The robot must return the milk carton and close the fridge. Leaving the fridge open means that the task was not fully completed.

Therefore, success must be evaluated over the entire sequence of actions rather than only by checking the final position of the mug. Intermediate events and the final state of the environment are also relevant.

2. **E2 · Constrained Action Designator & Grounding**

    - Distinguish interpretation, inferred means, and grounded objects for this episode by filling out the three missing slots in the schematic action designator:
    
    - **Working Interpretation:** `(:type serving) (:recipient Eva) (:substance ________)`
    - **Inferred Function/Means:** `(:vessel-role ________)`

Solution: The initial action designator can be represented as:

Inferred Function/Means: (:vessel-role container-for-serving)

Grounded physical object: (:object Eva's mug on the counter)

The intended substance belongs to the interpretation of the request, while the vessel role is inferred as a necessary means of serving the drink. However, the type of milk (whole milk or oat drink) is still unclear and should be confirmed before pouring.


3. **E3 · The Commitment Decision & Asymmetric Failure Costs**

    - During the retrieval step, the robot faces a choice between a **5 mm test lift near the shelf** or a **high carry directly to the counter**. State the predicted effect, the cost of failure, and the single piece of evidence that justifies choosing the low test lift before committing to the carry.

Solution: The robot should perform a short test lift of approximately 5 mm rather than immediately carrying the carton to a higher position. As the surface is wet, the friction between the gripper and the carton may be insufficient to maintain a stable grasp. A short test lift allows the robot to check whether the carton slips while it is still close to the shelf. In case the carton slips, the robot can set it down and revise its grasp. In contrast, a slip during a high carry could cause the carton to fall, resulting in spillage or damage.
Therefore, the short test lift is a safer choice because it provides additional evidence about grasp stability while keeping the possible consequences of failure small.

4. **E4 · Causal Prediction across Material Variants**

    - Compare pouring milk into coffee versus pancake batter into a pan. Formulate the stopping rule for each liquid using `(:stop-when ...)`, and explain how material dynamics alter when the robot must command the wrist to tilt back.

Solution: 1. Pouring milk into coffee : The robot should target approximately 30 mL of milk and stop the pouring action before the full target amount is reached because some liquid may continue flowing after the carton is tilted back.

The action designator can be represented as:(:amount (some substance (:volume 30 mL)))(:stop-when (>= poured-volume 26.5 mL))(:monitor wrist-load)

The robot should monitor the wrist-load and the pouring process to estimate the amount transferred. The illustrative stopping threshold of 26.5 mL accounts for the milk that may still be in flight after the pouring motion stops.

2. Pouring pancake batter : For pancake batter, the robot should monitor the growing edge of the batter as it spreads across the pan. It should stop pouring before the batter reaches the desired final diameter because the batter continues spreading after the pour ends. For example, if the desired final diameter is approximately 12 cm, the robot could stop when the observed diameter approaches 11 cm and allow the batter to spread further. The main difference is that milk pouring is controlled primarily by the amount of liquid transferred, whereas batter pouring is controlled by monitoring the resulting shape and diameter.


5. **E5 · System Decomposition & The Reasoning Loop**

    - Trace the carton slip event through the seven-stage reasoning loop: **Infer → Ground → Predict → Decide → Regulate → Detect & Revise → Verify**. State precisely which step was triggered when downward motion was observed during the 5 mm lift, and how the action designator was developed in response.

Solution: The robot should follow an iterative reasoning process:

Infer: Determine that the milk carton needs to be retrieved and identify a suitable grasping action.

Ground: Identify the physical carton and recognise that its surface is wet.

Predict: Estimate that the side pinch grasp may not provide sufficient friction.

Decide: Choose a short test lift of approximately 5 mm to check grasp stability.

Regulate: Monitor the carton during the lift and detect any downward movement or slipping.

Detect & revise: If slipping occurs, update the belief that the current grasp is unreliable and revise the action designator to use bottom support instead.

Verify: Check that the carton is stable with the revised grasp before continuing with the task.

These steps represent different levels of reasoning. Updating the belief changes what the robot thinks about the situation. Revising the action designator changes the planned action. Changing the physical action means actually releasing or repositioning the gripper and transitioning to the new grasp. The robot must also consider whether bottom support allows sufficient wrist rotation for pouring. If not, it should plan a stable regrasp before starting the pouring action.


6. **E6 · Conceptual Ablation & Evidence of Competence**

    - Perform an ablation experiment. Suppose we remove the **"Check before commitment"** capability and replace it with a fixed default high-side-pinch lift. Predict a specific kitchen environment variant where this ablated system fails, and explain why single-run execution success differs from evidence of general competence across task variants.

Solution: An ablation experiment could be performed by removing the checking and revision steps from the robot's reasoning process. Instead of performing a test lift and checking for slipping, the robot would directly use a fixed high side-pinch grasp to lift the carton. With a dry carton, this approach might still work and succeed because the grasp could provide sufficient friction. However, with a wet carton, the grasp may become unstable and the carton may slip or fall, causing milk to spill. This experiment would demonstrate the importance of checking the outcome of an action and revising the plan when new evidence becomes available. However, success or failure in a single experiment is not sufficient to establish general competence. The robot should be tested across different surface conditions, carton weights, grasping situations and environments to evaluate whether it can handle a wider range of situations reliably.
